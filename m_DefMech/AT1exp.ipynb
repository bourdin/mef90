```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import scipy
```

```python
def a(s,b):
    if b == 0:
        return 1-s
    else:
        return 1 + (np.exp(-b*s) - 1) / (1 - np.exp(-b))
def ap0(b):
    if b == 0:
        return 1
    else:
        return -b / (1-np.e**(-b))
def sigmac(b):
    return np.sqrt(-2./ap0(b))

def cw(b):
    return scipy.integrate.quad(lambda s : np.sqrt(a(s,b)), 0, 1)[0]

b = np.linspace(0., 5., 11)
plt.close('all')
fig, ax = plt.subplots()
ax.plot(b, [cw(bb) for bb in b])
```

```python
N = 101
S = np.linspace(0,1,N)
for b in np.linspace(0,4,5):
    plt.plot(S, [a(s,b) for s in S],label = "b = {}".format(b))
plt.legend(loc=0)
plt.show()
```

```python
B = np.linspace(0.01,10,11)
plt.plot(B,[sigmac(b) for b in B])
plt.grid()
plt.show()
```
